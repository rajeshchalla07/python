```python
# sets
s={1,2,3,5}
s1=set()
s2={}
print(type(s),type(s1),type(s2))
```

```python
# add(element)
s.add(20)
print(s)
```

```python
# remove(element)
s.remove(5)
print(s) 
```

```python
s.remove(10)
```

```python
# discard(element)
s.discard(5)
print(s)
s.discard(10)
print(s)
```

```python
# set operations
s1={10,20,30,40}
s2={10,20,50,70}
print(s1)
print(s2)
# union
print(s1.union(s2))
print(s1|s2)
```

```python
# intersection
print(s1.intersection(s2))
print(s1&s2)
```

```python
# difference
print(s1.difference(s2))
print(s1-s2)
```

```python
# symmetric_difference()
print(s1.symmetric_difference(s2))
print(s1^s2)
```

```python
s3={22,33,44,55}
s4={22,33}
# issuperset()
# issubset()
print(s3.issuperset(s4))
print(s3.issubset(s4))
```

```python
# update() 
d2.update({'name':'Prabhas'})
print(d2)
d2.update(marks=6)
print(d2)
```
